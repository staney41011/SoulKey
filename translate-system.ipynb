```python
import subprocess
from pathlib import Path

repo = Path("/kaggle/working/SoulKey")

# 確保 repo 存在
if not (repo / ".git").exists():
    subprocess.run([
        "git", "clone",
        "https://github.com/staney41011/SoulKey.git",
        str(repo)
    ], check=True)

# 強制同步 GitHub main 最新版本
subprocess.run([
    "git", "-C", str(repo),
    "fetch", "origin", "main"
], check=True)

subprocess.run([
    "git", "-C", str(repo),
    "reset", "--hard", "origin/main"
], check=True)

print("\n=== 目前 Git commit ===")
subprocess.run([
    "git", "-C", str(repo),
    "log", "-1", "--oneline"
], check=True)

print("\n=== translation-system 內容 ===")
subprocess.run([
    "ls", "-la",
    str(repo / "translation-system")
], check=True)

script = repo / "translation-system" / "prepare_model.py"

print("\nprepare_model.py =", script)
print("是否存在 =", script.exists())

assert script.exists(), f"仍找不到：{script}"

print("\n✅ GitHub 程式已同步完成")
```
